```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import datetime
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, TensorBoard
from tensorflow.keras import layers, models, Input, regularizers
from tensorflow.keras.models import Sequential
from tensorflow.keras.applications import DenseNet121, ResNet50

DATASET_DIR = "C:\\Users\\ramana777\\Downloads\\CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42
EPOCHS = 10
LOGS_DIR = "logs_kidney"
os.makedirs(LOGS_DIR, exist_ok=True)

def create_dataframe(dataset_dir):
    filepaths, labels = [], []
    for cls in sorted(os.listdir(dataset_dir)):
        cls_path = os.path.join(dataset_dir, cls)
        if not os.path.isdir(cls_path):
            continue
        for img in os.listdir(cls_path):
            if img.lower().endswith(('.jpg', '.png', '.jpeg')):
                filepaths.append(os.path.join(cls_path, img))
                labels.append(cls)
    return pd.DataFrame({"filepath": filepaths, "labels": labels})

def split_dataframe(df, test_size=0.2, val_size=0.5, seed=SEED):
    train_df, dummy_df = train_test_split(df, test_size=test_size, random_state=seed, stratify=df["labels"])
    valid_df, test_df = train_test_split(dummy_df, test_size=val_size, random_state=seed, stratify=dummy_df["labels"])
    return train_df, valid_df, test_df

def build_generators(train_df, valid_df, test_df, img_size=IMG_SIZE, batch_size=BATCH_SIZE):
    train_datagen = ImageDataGenerator(
        rescale=1./255,
        rotation_range=15,
        zoom_range=0.15,
        horizontal_flip=True,
    )
    valid_datagen = ImageDataGenerator(rescale=1./255)
    test_datagen = ImageDataGenerator(rescale=1./255)

    train_gen = train_datagen.flow_from_dataframe(train_df, x_col="filepath", y_col="labels",
                                                  target_size=img_size, class_mode="categorical",
                                                  shuffle=True, batch_size=batch_size, seed=SEED)
    valid_gen = valid_datagen.flow_from_dataframe(valid_df, x_col="filepath", y_col="labels",
                                                  target_size=img_size, class_mode="categorical",
                                                  shuffle=True, batch_size=batch_size, seed=SEED)
    test_gen = test_datagen.flow_from_dataframe(test_df, x_col="filepath", y_col="labels",
                                                target_size=img_size, class_mode="categorical",
                                                shuffle=False, batch_size=batch_size)
    return train_gen, valid_gen, test_gen

def build_simple_cnn(input_shape=(224,224,3), num_classes=4):
    model = Sequential([
        layers.Conv2D(32, (3,3), activation="relu", input_shape=input_shape),
        layers.MaxPooling2D(2,2),
        layers.Conv2D(64, (3,3), activation="relu"),
        layers.MaxPooling2D(2,2),
        layers.Flatten(),
        layers.Dense(128, activation="relu"),
        layers.Dense(num_classes, activation="softmax")
    ])
    return model

def build_resnet50(input_shape=(224,224,3), num_classes=4):
    base = ResNet50(weights=None, include_top=False, input_shape=input_shape)
    base.load_weights("C:\\Users\\ramana777\\Downloads\\RadImageNet_models\\RadImageNet-ResNet50_notop.h5")
    base.trainable = False
    inputs = Input(shape=input_shape)
    x = tf.keras.applications.efficientnet_v2.preprocess_input(inputs)
    x = base(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dense(256, activation="relu", kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.Dropout(0.4)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)
    model = models.Model(inputs, outputs, name="ResNet50")
    return model

def build_densenet121(input_shape=(224,224,3), num_classes=4):
    base = DenseNet121(weights=None, include_top=False, input_shape=input_shape)
    base.load_weights("C:\\Users\\ramana777\\Downloads\\RadImageNet_models\\RadImageNet-DenseNet121_notop.h5")
    base.trainable = False
    inputs = Input(shape=input_shape)
    x = tf.keras.applications.densenet.preprocess_input(inputs)
    x = base(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dense(256, activation="relu", kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.Dropout(0.4)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)
    model = models.Model(inputs, outputs, name="DenseNet121")
    return model

def make_callbacks(name):
    timestamp = datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
    log_dir = os.path.join(LOGS_DIR, f"{name}_{timestamp}")
    tb = TensorBoard(log_dir=log_dir, histogram_freq=1)
    ckpt = ModelCheckpoint(f"{name.lower()}_best.h5", monitor="val_loss", save_best_only=True, verbose=1)
    es = EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True, verbose=1)
    return [tb, ckpt, es]

def plot_history(history, name):
    plt.figure(figsize=(12,5))
    plt.subplot(1,2,1)
    plt.plot(history.history.get("loss", []), label="train_loss")
    plt.plot(history.history.get("val_loss", []), label="val_loss")
    plt.title(f"{name} Loss"); plt.legend()
    plt.subplot(1,2,2)
    plt.plot(history.history.get("accuracy", []), label="train_acc")
    plt.plot(history.history.get("val_accuracy", []), label="val_acc")
    plt.title(f"{name} Accuracy"); plt.legend()
    plt.tight_layout(); plt.show()

def evaluate_and_report(model, test_gen, class_labels, name, show_n_images=6):
    preds_prob = model.predict(test_gen, verbose=0)
    preds = np.argmax(preds_prob, axis=1)
    y_true = test_gen.classes

    cm = confusion_matrix(y_true, preds)
    plt.figure(figsize=(6,5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=class_labels, yticklabels=class_labels)
    plt.title(f"{name} Confusion Matrix"); plt.xlabel("Predicted"); plt.ylabel("True")
    plt.show()

    print(f"\n{name} Classification Report:\n")
    print(classification_report(y_true, preds, target_names=class_labels))

    loss, acc = model.evaluate(test_gen, verbose=0)
    print(f"\n{name} Test Accuracy: {acc*100:.2f}% | Loss: {loss:.4f}")

    plt.figure(figsize=(15, 6))
    for i, idx in enumerate(np.random.choice(len(y_true), min(show_n_images, len(y_true)), replace=False)):
        img_path = test_gen.filepaths[idx]
        img = tf.keras.preprocessing.image.load_img(img_path, target_size=test_gen.target_size)
        arr = tf.keras.preprocessing.image.img_to_array(img) / 255.0
        pred_idx = preds[idx]
        pred_label = class_labels[pred_idx]
        conf = preds_prob[idx, pred_idx]
        true_label = class_labels[y_true[idx]]
        plt.subplot(2, 3, i+1)
        plt.imshow(arr); plt.axis("off")
        plt.title(f"True: {true_label}\nPred: {pred_label} ({conf*100:.1f}%)")
    plt.suptitle(f"{name} — Predictions"); plt.tight_layout(); plt.show()

    model.save(f"{name.lower()}_ctimages_kidney.h5")
    print(f"{name} saved as {name.lower()}_ctimages_kidney.h5")
    return acc

def train_and_run_all():
    df = create_dataframe(DATASET_DIR)
    print(f"Total images: {len(df)}")
    train_df, valid_df, test_df = split_dataframe(df)
    train_gen, valid_gen, test_gen = build_generators(train_df, valid_df, test_df)
    class_labels = list(train_gen.class_indices.keys())
    results = {}

    model_name = "ResNet50"
    model = build_resnet50(num_classes=len(class_labels))
    model.summary()
    cbs = make_callbacks(model_name)
    model.compile(optimizer=Adam(1e-4), loss="categorical_crossentropy", metrics=["accuracy"])
    print(f"\n--- Training {model_name} ---")
    hist = model.fit(train_gen, validation_data=valid_gen, epochs=EPOCHS, callbacks=cbs, verbose=1)
    plot_history(hist, model_name)
    results[model_name] = evaluate_and_report(model, test_gen, class_labels, model_name)

    model_name = "DenseNet121"
    model = build_densenet121(num_classes=len(class_labels))
    model.summary()
    cbs = make_callbacks(model_name)
    model.compile(optimizer=Adam(1e-4), loss="categorical_crossentropy", metrics=["accuracy"])
    print(f"\n--- Training {model_name} ---")
    hist = model.fit(train_gen, validation_data=valid_gen, epochs=EPOCHS, callbacks=cbs, verbose=1)
    plot_history(hist, model_name)
    results[model_name] = evaluate_and_report(model, test_gen, class_labels, model_name)

    print("\nFinal Test Accuracy Summary:")
    for k, v in results.items():
        print(f"{k}: {v*100:.2f}%")

    plt.figure(figsize=(8,5))
    plt.bar(results.keys(), [v*100 for v in results.values()], color=["#007acc", "#f28e2b", "#76b7b2"])
    plt.ylabel("Test Accuracy (%)")
    plt.title("Model Performance Comparison")
    plt.xticks(rotation=20)
    for i, v in enumerate(results.values()):
        plt.text(i, v*100 + 0.5, f"{v*100:.2f}%", ha="center", fontweight="bold")
    plt.tight_layout()
    plt.show()

    print(f"\nTensorBoard logs saved in: {LOGS_DIR}")
    print(f"Run in terminal: tensorboard --logdir={LOGS_DIR}")

if __name__ == "__main__":
    train_and_run_all()
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os
import warnings
warnings.filterwarnings("ignore")

IMG_PATHS = [
    r"C:\Users\ramana777\OneDrive\Desktop\kidney_stone.jpg",
    r"C:\Users\ramana777\OneDrive\Desktop\kidney.jpg",
    r"C:\Users\ramana777\OneDrive\Desktop\kidney1.jpg",
]

MODEL_FILES = [
    "simplecnn_kidney.h5",
    "densenet121_ctimages_kidney.h5",
    "resnet50_ctimages_kidney.h5",
]

CLASS_NAMES = ["Cyst", "Normal", "Stone", "Tumor"]
IMG_SIZE = (224, 224)

custom_objects = {
    "TrueDivide": tf.keras.layers.Lambda(lambda x: x),
    "tf": tf  
}

def predict_image(model, img):
    img = tf.image.resize(img, IMG_SIZE) / 255.0
    img = np.expand_dims(img, axis=0)
    pred = model.predict(img, verbose=0)
    return CLASS_NAMES[np.argmax(pred)], float(np.max(pred))

for img_path in IMG_PATHS:
    if not os.path.exists(img_path):
        print(f"⚠️ Image not found: {img_path}")
        continue

    img = mpimg.imread(img_path)
    plt.imshow(img)
    plt.title(f"Input Image: {os.path.basename(img_path)}")
    plt.axis("off")
    plt.show()

    if img.ndim == 2:
        img = np.stack([img]*3, axis=-1)
    elif img.shape[-1] == 4:
        img = img[..., :3]

    print(f"\n🩻 Predictions for: {os.path.basename(img_path)}")
    
    for model_file in MODEL_FILES:
        if not os.path.exists(model_file):
            print(f"⚠️ Model not found: {model_file}")
            continue

        try:
            model = tf.keras.models.load_model(model_file, custom_objects=custom_objects)
        except Exception as e:
            print(f"❌ Error loading {model_file}: {e}")
            continue

        try:
            label, conf = predict_image(model, img)
            print(f"   📦 {os.path.basename(model_file):<35} → {label:<10} ({conf:.2f})")
        except Exception as e:
            print(f"❌ Prediction failed for {model_file}: {e}")

    print("-" * 60)
```
